```
from transformers import pipeline

gen = pipeline("text-generation", model="gpt2")

out = gen(
    "I'm very happy to study here",
    max_length=20,
    num_return_sequences=2
)

n = len(out)
for i in range(n):
    data = out[i]
    txt = data["generated_text"]
    print(f"第{i+1}段：{txt}")
```
