```python
from transformers import pipeline

# 零样本分类，和你之前写的一样
classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")

text = "我去你真的是人类呀"
labels = ["赞扬", "抱怨", "震惊", "调侃"]

# 运行，得到 result（和你写的一样）
result = classifier(text, labels)

# 打印完整 result，看看它长什么样
print(result)
```
